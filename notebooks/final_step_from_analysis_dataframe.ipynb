# Final step from `analysis_ready_dataframe.csv`

This notebook demonstrates how to load the processed dataframe exported by the main pipeline. Use this notebook for final plots, dose comparisons and publication-style figures.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

ANALYSIS_READY_CSV = "/content/spectro_single_colab_pipeline/results/processed/analysis_ready_dataframe.csv"
df = pd.read_csv(ANALYSIS_READY_CSV)
display(df.head())
print(df["analysis_type"].value_counts())
```

## Example: plot same-current transmittance curves

```python
trans = df[df["analysis_type"] == "same_current_transmittance"].copy()
if trans.empty:
    print("No same-current transmittance records found.")
else:
    plt.figure(figsize=(12,7))
    for label, g in trans.groupby(["sample_id", "dose_label", "comparison_id"]):
        plt.plot(g["wavelength_nm"], g["transmittance_smooth"], label=" | ".join(map(str, label)), linewidth=2)
    plt.xlabel("Wavelength [nm]")
    plt.ylabel("Transmittance")
    plt.title("Final-step transmittance plot")
    plt.grid(alpha=0.25)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()
```

## Example: load integral summary

```python
integrals_path = Path(ANALYSIS_READY_CSV).parents[1] / "tables" / "integral_summary.csv"
if integrals_path.exists():
    integrals = pd.read_csv(integrals_path)
    display(integrals)
else:
    print("Integral summary not found:", integrals_path)
```
